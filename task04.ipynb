## 4.2 Kinematics of a downstream segment
Assume you know the global state of a parent segment and the relative motion to a downstream segment.

A typical recursive formulation contains terms like:
$$r_c = r_p + R_p \, r_{pc}$$
$$v_c = v_p + \omega_p \times (R_p \, r_{pc}) + v_{rel}$$
$$a_c = a_p + \alpha_p \times (R_p \, r_{pc}) + \omega_p \times (\omega_p \times (R_p \, r_{pc})) + a_{rel}$$

where `p` is the parent segment and `c` is the child segment.
Your exact expressions depend on whether the joint is a hinge or a slider.

```python
import numpy as np

omega = np.array([0.0, 0.0, 2.0])
alpha = np.array([0.0, 0.0, 0.5])
r = np.array([0.4, 0.0, 0.0])

v = np.cross(omega, r)
a = np.cross(alpha, r) + np.cross(omega, np.cross(omega, r))
print('Tangential / rotational velocity:', v)
print('Tangential + centripetal acceleration:', a)
```

## 4.3 Load `q`, `qdot` and `qddot`
We start from the cleaned joint-angle trajectories and approximate first and second derivatives numerically.

```python
import pandas as pd
from model.kintree import get_model_dictionary

q_all = pd.read_csv('data/angles_clean.csv')
model_dict = get_model_dictionary()
key = list(q_all.columns[1:])
time = q_all['time'].to_numpy()
q = q_all[key].to_numpy()
qdot = np.gradient(q, time, axis=0)
qddot = np.gradient(qdot, time, axis=0)

frame = 100
print('q shape:', q.shape)
print('qdot shape:', qdot.shape)
print('qddot shape:', qddot.shape)
```
